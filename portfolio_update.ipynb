```python
from pathlib import Path
import json
ROOT = Path.cwd()
content_path = ROOT / 'site_content.json'
content = json.loads(content_path.read_text(encoding='utf-8'))

# Edit these sample values. Replace them with your own information.
content['name'] = 'Michael J Bailey'
content['role'] = 'Developer'
content['tagline'] = 'Thoughtful ideas. Useful things.'
content['intro'] = 'Write a short introduction about what you make and what interests you.'
content['email'] = 'your-real-email@example.com'
content['about'] = ['Write a paragraph about yourself.', 'Add what you enjoy building, learning, or exploring.']
content['interests'] = ['Web development', 'Creative problem-solving', 'Learning by making']
content['projects'] = [
    {'title': 'Project title', 'category': 'Web development', 'description': 'What it does and why it matters.', 'url': 'https://example.com', 'year': '2026', 'details': ['What you built and the problem it solves.', 'Tools, decisions, or lessons worth sharing.']}
]
content['other_work'] = [{'title': 'Experiments & side projects', 'description': 'Small builds and ideas I am exploring.', 'year': 'Ongoing'}]
content['notes'] = [{'title': 'A note title', 'date': '2026', 'summary': 'A short preview of your note.'}]
content['social'] = [{'label': 'GitHub', 'url': 'https://github.com/your-user'}]
content['contact_copy'] = 'Have a question or want to talk about a project? I would be glad to hear from you.'
content_path.write_text(json.dumps(content, indent=2, ensure_ascii=False) + '\n', encoding='utf-8')
print('Saved', content_path)
```
